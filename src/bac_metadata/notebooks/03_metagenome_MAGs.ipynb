```python
import pandas as pd


path = "/Users/davidabelson/Library/CloudStorage/OneDrive-UniversityofCambridge/Code_repos/md_curation/41586_2019_965_MOESM3_ESM.xlsx"
# Read the excel file, with headings in the first row
df = pd.read_excel(path, header=1)
df.head()
```

```python
# Rename "Disease name1,2" to "disease_name; Antibiotics3 to antibiotics; "
df.rename(columns={"Health state": "health_state", 
      "Disease name1,2": "disease_name", 
      "Antibiotics3": "antibiotics", 
      "Age group4": "age_group", 
      "Location": "location",
      "Benchmarked5": "benchmarked"},
      inplace=True
      )
df.head()

# # Count categories in healthy_state, disease_name, antibiotics, age_group, benchmarked
print(df["health_state"].value_counts())
print(df["disease_name"].value_counts())
print(df["antibiotics"].value_counts())
print(df["age_group"].value_counts())
print(df["location"].value_counts())
print(df["benchmarked"].value_counts())

# Count number of rows in df
```

```python
near_complete_geneomes = "/Users/davidabelson/Library/CloudStorage/OneDrive-UniversityofCambridge/Code_repos/md_curation/41586_2019_965_MOESM5_ESM.xlsx"
near_complete_geneomes_df = pd.read_excel(near_complete_geneomes, header=1)
near_complete_geneomes_df.head()

# How many of near_complete_geneomes_df["Metagenome1"] are in df["Run"]?
print(near_complete_geneomes_df["Metagenome1"].isin(df["Run"]).value_counts())

# Number of rows per metagenome - histogram with 25 bins
plot = near_complete_geneomes_df["Metagenome1"].value_counts().hist(bins=25)
# Add title and axis labels
plot.set_title("MAGs per Stool Sample")
plot.set_xlabel("Number of MAGs / Stool Sample")
plot.set_ylabel("Number of Stool Samples")

# Show plot
import matplotlib.pyplot as plt
plt.show()
```
