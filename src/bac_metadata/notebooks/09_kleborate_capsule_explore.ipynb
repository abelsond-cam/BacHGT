```python
# Restrict ourselves to kpsc samples
kpsc_metadata = metadata[metadata['is_kpsc']]
# Number of kpsc samples
print(f"Number of kpsc samples: {len(kpsc_metadata)}")
# Number of unique countries in kpsc samples
print(f"Number of unique countries in kpsc samples: {kpsc_metadata['country'].nunique()}")
# Number of unique species in kpsc samples
print(f"Number of unique species in kpsc samples: {kpsc_metadata['species'].nunique()}")
```

```python
# Top 5 Sublieage
top_5_sublineages = kpsc_metadata['Sublineage'].value_counts().head(5)
print(top_5_sublineages)

# Look at the ST's in the top 5 sublineages
for sublineage in top_5_sublineages.index:
    print("-"*100)
    print(f"Sublineage: {sublineage}")
    #Print the number of samples in the sublineage
    cg_in_sublineage = kpsc_metadata[kpsc_metadata['Sublineage']==sublineage]['Clonal group'].value_counts()
    # Print the number of sublineages 
    print(f"Number of clonal groups: {len(cg_in_sublineage)}")
    # Print sublineages with > 100 samples
    print("Clonal groups with > 100 samples:")
    print(cg_in_sublineage[cg_in_sublineage > 100])
```

```python
# Print the % number of "-" values of the K_Missing_expected_genes and O_Missing_expected_genes
k_missing = (kpsc_metadata['K_Missing_expected_genes'] == "-").sum()
prop_k_loci_error = 1- k_missing/len(kpsc_metadata)
print(f"% with missing genes in K_Missing_expected_genes: {prop_k_loci_error*100:.2f}%")
print(f"Equates to the number of missing values of K_Missing_expected_genes: {prop_k_loci_error*len(kpsc_metadata):.0f}")
# Count number of k_locus containing 'null'
num_null_k_type = kpsc_metadata['K_type'].str.contains('null', na=False).sum()
print(f"Number of K_locus containing 'null': {num_null_k_type}")

o_missing = (kpsc_metadata['O_Missing_expected_genes'] == "-").sum()
print(f"% missing values of O_Missing_expected_genes: {o_missing/len(kpsc_metadata)*100:.2f}%")
```
